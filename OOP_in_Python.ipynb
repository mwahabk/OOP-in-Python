```python
class Practice():
  def __init__ (self,name, marks):
    self.name = name
    self.marks = marks

  def get_avg(self):
    sum = 0
    for i in self.marks:
      sum += i
    print(f' your avg score is:', sum/3)

S1 = Practice ("Wahab", [99,88,90] )
S1.get_avg()
S2 = Practice ("Ali", [100,80,90] )
S2.get_avg()
S3 = Practice ("Saad", [90,80,90] )
S3.get_avg()
S4 = Practice ("Ahmad", [60,80,90] )
S4.get_avg()
S5 = Practice ("Zaid", [99,80,90] )
S5.get_avg()
```

## 4 Building Blocks of OOP

**Abstraction** :
Hides complex details and shows only essential features.

**Encapsulation:** Encapsulation is about hiding internal details.

public and Private attributes & methods are meant to be used only within the class and are not accessible from outside the class.



**Inheritance:** Inheritance is when one class gets the properties and methods of another class.

It’s like a child class inheriting from a parent class.



**Ploymorphism :** means "many forms".

It allows different classes to have methods with the same name, but behave differently based on the object.

## Practice

Create Account class with 2 attributes – balance & account no.
Create methods for debit, credit & printing the balance.



```python
class Account():
    def __init__(self, bal, acc):
        self.balance = bal
        self.account_no = acc
        print("total balance = " , self.get_balance())

    def debit(self, amount):
        self.balance -= amount
        print(f"Rs. {amount} was debited")
        print("total balance = " , self.get_balance())

    def credit(self, amount):
        self.balance += amount
        print(f"Rs. {amount} was credited")
        print("total balance = " , self.get_balance())

    def get_balance(self):
        return self.balance

# Create object and use methods (outside the class, no indentation here)
Account1 = Account(5000, 1234)
Account1.debit(1000)
Account1.credit(2000)
```

## Encapsulation

```python
class Account:
    def __init__(self, acc_no, acc_pass):
        self.acc_no = acc_no
        self.__acc_pass = acc_pass  # Private attribute

    def reset_pass(self):
        print(self.__acc_pass)


acc1 = Account("12345", "abcde")

print(acc1.acc_no)         # This will work
print(acc1.__acc_pass)     # This will raise an AttributeError
```

## Inheritance

```python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"My name is {self.name}.")

class Student(Person):
    def __init__(self, name, roll_no):
        Person.__init__(self, name)   # Directly calling parent class constructor
        self.roll_no = roll_no

    def show_roll(self):
        print(f"My roll number is {self.roll_no}.")

# Creating an object of Student
s1 = Student("Wahab", 102)
s1.introduce()     # From Person class
s1.show_roll()     # From Student class

```

## Polymorphism

```python
class Animal:
    def make_sound(self):
        print("Some generic animal sound")

class Dog(Animal):
    def make_sound(self):
        print("Woof!")

class Cat(Animal):
    def make_sound(self):
        print("Meow!")

# Polymorphism in action
animals = [Dog(), Cat(), Animal()]

for animal in animals:
    animal.make_sound()
```
